# 🦘 CycloneX — Kangaroo Solver (Kaggle Standalone Edition)

Executa o **Kangaroo Solver** de alta performance na GPU do Kaggle sem depender de conexões de rede ou servidor Master.

## 🚀 Instruções de Uso:
1. No painel à direita do Kaggle, ative **Accelerator → GPU T4 x2** (ou GPU P100).
2. Ative **Internet → On** (obrigatório para clonar o repositório).
3. Defina os parâmetros do Puzzle na **Célula 4** (Puzzle #140 vem pré-configurado).
4. Clique em **Run All** no topo da página.

> 💾 **Garantia de Resultado:** Se a chave privada for encontrada, ela é gravada imediatamente em `/kaggle/working/FOUND_KEY.txt` (arquivo persistente que fica salvo na aba *Output* do Kaggle para download).

## ⚙️ Célula 1 — Verificar GPUs no Kaggle

```python
import subprocess, os, sys, time, datetime

print('=== INFORMAÇÕES DA GPU ===')
result = subprocess.run(['nvidia-smi'], capture_output=True, text=True)
if result.returncode != 0:
    raise RuntimeError('❌ GPU não detectada! Ative GPU T4 x2 no painel direito (Accelerator).')
print(result.stdout)

cc_result = subprocess.run(
    ['nvidia-smi', '--query-gpu=index,name,compute_cap,memory.total', '--format=csv,noheader'],
    capture_output=True, text=True
)
gpus = [g.strip() for g in cc_result.stdout.strip().split('\n') if g.strip()]
print(f'\n✅ {len(gpus)} GPU(s) detectada(s):')
for g in gpus:
    print(f'   {g}')

NUM_GPUS = len(gpus)
```

## 📥 Célula 2 — Clonar Repositório CycloneX

```python
REPO_DIR = '/kaggle/working/cycloneX'

if os.path.exists(REPO_DIR):
    print('Atualizando repositório existente...')
    subprocess.run(['git', '-C', REPO_DIR, 'pull'], capture_output=True, text=True)
else:
    print('Clonando repositório CycloneX...')
    res = subprocess.run(['git', 'clone', 'https://github.com/ggsofthouse/cycloneX.git', REPO_DIR], capture_output=True, text=True)
    if res.returncode != 0:
        raise RuntimeError(f'❌ Falha ao clonar repo: {res.stderr}\nVerifique se Internet está ON no painel direito.')

os.chdir(REPO_DIR)
print(f'✅ Repositório pronto em: {REPO_DIR}')
```

## 🔨 Célula 3 — Compilar CUDACyclone (Linux / CUDA 12)

```python
CUDA_DIR = f'{REPO_DIR}/CUDACyclone-main'
BINARY   = f'{CUDA_DIR}/CUDACyclone'

# Garantir nvcc no PATH
for cuda_path in ['/usr/local/cuda/bin', '/usr/local/cuda-12/bin', '/usr/local/cuda-11/bin']:
    if os.path.exists(f'{cuda_path}/nvcc'):
        os.environ['PATH'] = f"{cuda_path}:{os.environ.get('PATH', '')}"
        break

cc_raw = subprocess.run(
    ['nvidia-smi', '--query-gpu=compute_cap', '--format=csv,noheader'],
    capture_output=True, text=True
).stdout.strip().split('\n')[0].replace('.', '')
GPU_ARCH = cc_raw.strip() if cc_raw.strip() else '75'
print(f'GPU Compute Capability: sm_{GPU_ARCH}')
print('Compilando CUDACyclone para GPU Linux (aguarde ~2 min)...')

subprocess.run(['make', 'clean'], cwd=CUDA_DIR, capture_output=True)
build = subprocess.run(['make', f'GPU_ARCHS={GPU_ARCH}', '-j4'], capture_output=True, text=True, cwd=CUDA_DIR)

if build.returncode != 0 or not os.path.exists(BINARY):
    raise RuntimeError('❌ Compilação falhou! Log:\n' + build.stderr[-2000:])

print(f'✅ Compilado com sucesso! Tamanho do binário: {os.path.getsize(BINARY)/(1024*1024):.1f} MB')
```

## 🎯 Célula 4 — Configurar o Alvo (Puzzle #140)

```python
# Configuração do Alvo
PUZZLE_NUM    = "140"
TARGET_PUBKEY = "031f6a332d3c5c4f2de2378c012f429cd109ba07d69690c6c701b6bb87860d6640"
TARGET_ADDR   = "1QKBaU6WAeycb3DbKbLBkX7vJiaS8r42Xo"
RANGE_STR     = "800000000000000000000000000000000000:ffffffffffffffffffffffffffffffffffff"
DP_BITS       = 24
GRID          = "512,1024"
SLICES        = 256

FOUND_FILE    = '/kaggle/working/FOUND_KEY.txt'
JSON_FILE     = '/kaggle/working/puzzles_solved.json'

print(f'Puzzle       : #{PUZZLE_NUM}')
print(f'Target Addr  : {TARGET_ADDR}')
print(f'Public Key   : {TARGET_PUBKEY}')
print(f'Range        : {RANGE_STR}')
print(f'DP Bits      : {DP_BITS}')
print(f'Arquivo final: {FOUND_FILE}')
```

## 🚀 Célula 5 — Executar Kangaroo Solver & Gravação Automática

O script executa o solver em cada GPU. Se a chave for encontrada, ela é gravada em `/kaggle/working/FOUND_KEY.txt` e exibida no painel em letras garrafais!

```python
import re, json

def save_found_key(priv_hex):
    now_str = datetime.datetime.now(datetime.timezone.utc).strftime('%Y-%m-%d %H:%M:%S UTC')
    content = f"""==============================================================
🎉🎉🎉 CHAVE PRIVADA ENCONTRADA! 🎉🎉🎉
Data/Hora : {now_str}
Puzzle    : #{PUZZLE_NUM}
Endereço  : {TARGET_ADDR}
Pubkey    : {TARGET_PUBKEY}
Chave (HEX): {priv_hex}
==============================================================
"""
    # 1. Gravar em FOUND_KEY.txt na pasta output do Kaggle
    with open(FOUND_FILE, 'w') as f:
        f.write(content)
        
    # 2. Gravar em JSON
    solved_data = {
        "puzzle": PUZZLE_NUM,
        "address": TARGET_ADDR,
        "pubkey": TARGET_PUBKEY,
        "private_key": priv_hex,
        "found_at": now_str
    }
    with open(JSON_FILE, 'w') as f:
        json.dump(solved_data, f, indent=2)
        
    print('\n' + '='*70)
    print(content)
    print(f'✅ Salvo com sucesso em: {FOUND_FILE}')
    print('='*70 + '\n')

print('🚀 Iniciando Kangaroo Solver nas GPUs do Kaggle...')
processes = []

# Executar uma instância por GPU disponível
for gpu_id in range(NUM_GPUS):
    cmd = [
        BINARY,
        '--solver', 'kangaroo',
        '--range', RANGE_STR,
        '--target-pubkey', TARGET_PUBKEY,
        '--address', TARGET_ADDR,
        '--dp-bits', str(DP_BITS),
        '--grid', GRID,
        '--slices', str(SLICES),
        '--gpus', str(gpu_id)
    ]
    env = {**os.environ, 'CUDA_VISIBLE_DEVICES': str(gpu_id)}
    proc = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1, env=env)
    processes.append((gpu_id, proc))
    print(f'   - GPU #{gpu_id} disparada...')

print('\n[Kangaroo Worker Kaggle] Processando... (Acompanhe a velocidade em tempo real below)\n')

try:
    found_key = None
    while any(p.poll() is None for _, p in processes):
        for gpu_id, p in processes:
            line = p.stdout.readline()
            if not line:
                continue
            line_str = line.strip()
            
            # Verificar se a chave foi encontrada
            if 'KEY FOUND' in line_str or 'Private key' in line_str:
                match = re.search(r'(?:Private key|KEY FOUND)[:\s]+([0-9A-Fa-fx]+)', line_str, re.IGNORECASE)
                if match:
                    found_key = match.group(1)
                    save_found_key(found_key)
                    # Encerrar todos os processos
                    for _, pr in processes:
                        pr.terminate()
                    break
                    
            if 'Speed:' in line_str or 'Time:' in line_str:
                sys.stdout.write(f'\r[GPU {gpu_id}] {line_str[:100]:<100}')
                sys.stdout.flush()
        time.sleep(0.01)
except KeyboardInterrupt:
    print('\n⛔ Interrompido pelo usuário.')
finally:
    for _, p in processes:
        if p.poll() is None:
            p.terminate()

if os.path.exists(FOUND_FILE):
    print('\n🎉 ATENÇÃO: CHAVE SALVA EM:', FOUND_FILE)
    with open(FOUND_FILE) as f:
        print(f.read())
else:
    print('\nSessão encerrada sem colisão encontrada neste ciclo.')
```
